Fine tune distilbert
Retrain and subset

```python
ft_distillbert_movie_details = pd.read_csv("./files/movie__details.csv")
ft_distillbert_reviews = pd.read_csv("./files/reviews.csv")

ft_distillbert_train_df, ft_distillbert_test_df = train_test_split(
    ft_distillbert_reviews,
    test_size=0.2,
    stratify=ft_distillbert_reviews['is_spoiler'],
    random_state=42
)
```

```python
print("Training set size:", len(ft_distillbert_train_df))
print("Validation set size:", len(ft_distillbert_test_df))
```

```python
class SpoilerDataset(Dataset):
    def __init__(self, reviews, labels, tokenizer, max_length=256):
        self.reviews = reviews
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.reviews)

    def __getitem__(self, idx):
        review = str(self.reviews[idx])
        encoding = self.tokenizer(
            review,
            truncation=True,
            padding='max_length',
            max_length=self.max_length,
            return_tensors='pt'
        )
        # Squeeze to remove extra batch dimension
        encoding = {key: val.squeeze(0) for key, val in encoding.items()}
        encoding['labels'] = torch.tensor(self.labels[idx], dtype=torch.long)
        return encoding

# Verify the Dataset by creating a small instance
model_name = "distilbert-base-uncased"
tokenizer = DistilBertTokenizerFast.from_pretrained(model_name)

sample_dataset = SpoilerDataset(ft_distillbert_train_df['review_text'].tolist(), ft_distillbert_train_df['is_spoiler'].tolist(), tokenizer)
print("Sample dataset length:", len(sample_dataset))
print("First sample:", sample_dataset[0])
```

```python
ft_model = DistilBertForSequenceClassification.from_pretrained(model_name, num_labels=2)
ft_distillbert_train_dataset = SpoilerDataset(ft_distillbert_train_df['review_text'].tolist(), ft_distillbert_train_df['is_spoiler'].tolist(), tokenizer)
ft_distillbert_val_dataset = SpoilerDataset(ft_distillbert_test_df['review_text'].tolist(), ft_distillbert_test_df['is_spoiler'].tolist(), tokenizer)
```

Set up trainer code

```python
training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=4,                   
    per_device_train_batch_size=16,       
    per_device_eval_batch_size=16,
    gradient_accumulation_steps=2,        
    evaluation_strategy="epoch",
    save_strategy="epoch",
    logging_dir="./logs",
    logging_steps=50,
    fp16=True,                           
    save_total_limit=2,
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss"
)

trainer = Trainer(
    model=ft_model,
    args=training_args,
    train_dataset=ft_distillbert_train_dataset,
    eval_dataset=ft_distillbert_val_dataset
)

trainer.train()
```

```python
eval_results = trainer.evaluate()
print("Evaluation Results:", eval_results)

model.save_pretrained("./fine_tuned_model")
tokenizer.save_pretrained("./fine_tuned_model")
```

```python
predictions_output = trainer.predict(ft_distillbert_val_dataset)


predicted_labels = np.argmax(predictions_output.predictions, axis=1)

true_labels = ft_distillbert_test_df['is_spoiler'].values

print(classification_report(true_labels, predicted_labels))
```
